```python
import gradio as gr
import pandas as pd
import random

# Mock Data for Movies
movies_db = {
    "Barbie": {
        "genre": "Comedy/Fantasy",
        "duration": "1h 54m",
        "rating": "PG-13",
        "image": "https://images.unsplash.com/photo-1594909122845-11baa439b7bf?auto=format&fit=crop&w=400&q=80",
        "times": ["12:00 PM", "3:00 PM", "6:00 PM", "9:00 PM"],
        "price": 12.50
    },
    "Legally Blonde": {
        "genre": "Comedy/Romance",
        "duration": "1h 36m",
        "rating": "PG-13",
        "image": "https://images.unsplash.com/photo-1489599849927-2ee91cede3ba?auto=format&fit=crop&w=400&q=80",
        "times": ["1:00 PM", "4:00 PM", "7:00 PM", "10:00 PM"],
        "price": 11.00
    },
    "Mean Girls": {
        "genre": "Comedy/Teen",
        "duration": "1h 37m",
        "rating": "PG-13",
        "image": "https://images.unsplash.com/photo-1517604931442-7e0c8ed2963c?auto=format&fit=crop&w=400&q=80",
        "times": ["11:30 AM", "2:30 PM", "5:30 PM", "8:30 PM"],
        "price": 12.00
    },
    "lil femboy": {
        "genre": "Drama/Comedy",
        "duration": "1h 45m",
        "rating": "R",
        "image": "https://images.unsplash.com/photo-1517604931442-7e0c8ed2963c?auto=format&fit=crop&w=400&q=80",
        "times": ["1:00 PM", "4:30 PM", "8:00 PM"],
        "price": 10.50
    }
}

# Dynamic seat map database to persist taken seats
# Structure: {(movie, showtime): [list of seat numbers that are taken]}
if 'seat_bookings_db' not in globals():
    seat_bookings_db = {}

# Custom ultra-pink & white theme config - Background changed to bright pink
pink_white_theme = gr.themes.Default(
    primary_hue="pink",
    secondary_hue="rose",
    neutral_hue="pink",
).set(
    body_background_fill="#FF69B4",  # Bright Hot Pink background
    body_background_fill_dark="#2D0B18",
    button_primary_background_fill="#FF1493",  # Deep Pink
    button_primary_background_fill_hover="#C71585",
    button_primary_text_color="#FFFFFF",
    block_background_fill="#FFFFFF",  # Pure White panels
    block_border_width="2px",
    block_border_color="#FFC0CB",  # Classic Pink border
    block_label_text_color="#D81B60",  # Dark Pink text
    input_background_fill="#FFFFFF",
    input_border_color="#FFC0CB",
    input_border_color_focus="#FF69B4"
)

# Custom CSS to grey out checked (selected) or taken options inside the CheckboxGroup
custom_css = """
footer {visibility: hidden !important;}

/* Styling checkbox items to look like ticket seats */
.gr-checkbox-group label {
    background-color: #FFF0F5 !important;
    border: 2px solid #FFC0CB !important;
    border-radius: 8px !important;
    padding: 8px 12px !important;
    transition: all 0.2s ease-in-out !important;
}

/* Grey out checked (selected) seats */
.gr-checkbox-group label.selected,
.gr-checkbox-group input:checked + span,
.gr-checkbox-group label:has(input:checked) {
    background-color: #E0E0E0 !important;
    border-color: #9E9E9E !important;
    color: #757575 !important;
    text-decoration: line-through !important;
    opacity: 0.8 !important;
}

/* Styling headings and standard items */
.title h1 {
    color: #FFFFFF !important; /* Changed to white for better contrast on hot pink background */
    text-shadow: 2px 2px 4px rgba(0, 0, 0, 0.4);
    font-weight: 800 !important;
}
"""

def get_seats_layout(movie_name, show_time):
    # Initialize dynamic storage for this session
    key = (movie_name, show_time)
    if key not in seat_bookings_db:
        # Generate a baseline of random taken seats
        rows = ['A', 'B', 'C', 'D']
        taken = []
        for r in rows:
            for c in range(1, 7):
                if random.random() < 0.25:
                    taken.append(f"{r}{c}")
        seat_bookings_db[key] = taken

    rows = ['A', 'B', 'C', 'D']
    seats = []
    for r in rows:
        for c in range(1, 7):
            seat_num = f"{r}{c}"
            status = "Taken" if seat_num in seat_bookings_db[key] else "Available"
            seats.append(f"{seat_num} ({status})")
    return seats

def update_movie_details(movie_name):
    movie = movies_db[movie_name]
    details = f"**Genre:** {movie['genre']} | **Duration:** {movie['duration']} | **Rating:** {movie['rating']} | **Ticket Price:** ${movie['price']:.2f}"
    times_choices = gr.Dropdown(choices=movie["times"], value=movie["times"], label="Select Show Time")
    seats_choices = gr.CheckboxGroup(choices=get_seats_layout(movie_name, movie["times"][0]), label="Select Seats (Checked/Taken seats will grey out!)", value=[])
    return details, times_choices, seats_choices

def update_seats_for_time(movie_name, show_time):
    return gr.CheckboxGroup(choices=get_seats_layout(movie_name, show_time), value=[])

def book_tickets(movie_name, show_time, selected_seats, customer_name, customer_email):
    if not selected_seats:
        return "❌ Please select at least one seat to book tickets.", gr.CheckboxGroup(choices=get_seats_layout(movie_name, show_time), value=[])
    if not customer_name or not customer_email:
        return "❌ Please fill in your name and email address.", gr.CheckboxGroup(choices=get_seats_layout(movie_name, show_time), value=selected_seats)

    taken_seats = [s for s in selected_seats if "Taken" in s]
    if taken_seats:
        return f"❌ Seats {', '.join([s.split()[0] for s in taken_seats])} are already taken. Please select available seats.", gr.CheckboxGroup(choices=get_seats_layout(movie_name, show_time), value=selected_seats)

    clean_seats = [s.split()[0] for s in selected_seats]

    # Lock the selected seats as taken in our database
    key = (movie_name, show_time)
    if key not in seat_bookings_db:
        seat_bookings_db[key] = []
    for seat in clean_seats:
        if seat not in seat_bookings_db[key]:
            seat_bookings_db[key].append(seat)

    price_per_ticket = movies_db[movie_name]["price"]
    total_price = len(clean_seats) * price_per_ticket
    booking_id = f"TKT-{random.randint(100000, 999999)}"

    receipt = f"""
    ┐───────────────────────────└
    🎫 **BOOKING CONFIRMED!** 🎫
    -------------------------------------
    **Booking ID:** {booking_id}
    **Customer:** {customer_name} ({customer_email})
    **Movie:** {movie_name}
    **Showtime:** {show_time}
    **Seats:** {', '.join(clean_seats)}
    -------------------------------------
    **Total Paid:** ${total_price:.2f}

    We look forward to seeing you at our pink paradise theater!
    """
    return receipt, gr.CheckboxGroup(choices=get_seats_layout(movie_name, show_time), value=[])

with gr.Blocks(elem_classes="title") as demo:
    gr.Markdown("# 🎬 Pink Cinema Booking Portal 🍿")

    # Displaying user's uploaded image as the official main poster
    gr.Image(
        value="/content/WhatsApp Image 2026-11-22 at 10.43.44 PM.jpeg",
        label="Cinema Main Poster",
        show_label=True,
        height=450
    )

    gr.Markdown("Choose a pink classic, select your seats, and prepare for a glamorous show!")

    with gr.Row():
        with gr.Column(scale=1):
            movie_selector = gr.Radio(
                choices=list(movies_db.keys()),
                value="Barbie",
                label="Choose Movie"
            )
            movie_info = gr.Markdown("**Genre:** Comedy/Fantasy | **Duration:** 1h 54m | **Rating:** PG-13 | **Ticket Price:** $12.50")
            showtime_dropdown = gr.Dropdown(choices=movies_db["Barbie"]["times"], value="12:00 PM", label="Select Show Time")

        with gr.Column(scale=2):
            gr.Markdown("### 💺 Screen Direction (Front of Room) 💺")
            seats_checkboxes = gr.CheckboxGroup(choices=get_seats_layout("Barbie", "12:00 PM"), label="Select Seats (Checked seats will automatically grey out)")

    with gr.Row():
        with gr.Column():
            gr.Markdown("### 👤 Customer Information")
            name_input = gr.Textbox(label="Full Name", placeholder="Elle Woods")
            email_input = gr.Textbox(label="Email Address", placeholder="elle@harvard.edu")

    book_button = gr.Button("💕 Confirm & Book Tickets", variant="primary")
    output_receipt = gr.Markdown()

    # Setup interactivity
    movie_selector.change(
        fn=update_movie_details,
        inputs=[movie_selector],
        outputs=[movie_info, showtime_dropdown, seats_checkboxes]
    )

    showtime_dropdown.change(
        fn=update_seats_for_time,
        inputs=[movie_selector, showtime_dropdown],
        outputs=[seats_checkboxes]
    )

    book_button.click(
        fn=book_tickets,
        inputs=[movie_selector, showtime_dropdown, seats_checkboxes, name_input, email_input],
        outputs=[output_receipt, seats_checkboxes]
    )

# Launching with updated bright pink theme
demo.launch(theme=pink_white_theme, css=custom_css, quiet=True, share=True)
```
